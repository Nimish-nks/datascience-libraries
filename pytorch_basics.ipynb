```python
import torch 
import numpy as np 
import pandas as pd

import matplotlib.pyplot as plt
```

```python
# Plot vecotrs, please keep the parameters in the same length
# @param: Vectors = [{"vector": vector variable, "name": name of vector, "color": color of the vector on diagram}]
%matplotlib widget
    
def plotVec(vectors):
    ax = plt.axes()
    
    # For loop to draw the vectors
    for vec in vectors:
        ax.arrow(0, 0, *vec["vector"], head_width = 0.05,color = vec["color"], head_length = 0.1)
        plt.text(*(vec["vector"] + 0.1), vec["name"])
    
    plt.ylim(-2,2)
    plt.xlim(-2,2)
```

#### Create tensor
- type
- datatype
- setting datatype manually
- size and dimension 

```python
_list=[1,2,3,4]
_tensor=torch.tensor(_list)
print(_tensor)
print(f'Type: {_tensor.type()}')
print(f'Type: {_tensor.dtype}')

# create float tensor
_tensor=torch.tensor([1.0,4.5])
print(_tensor)
print(f'Type: {_tensor.type()}')
print(f'Type: {_tensor.dtype}')

_tensor=torch.tensor([1.0,4.5],dtype=torch.float64)
print(f'Type: {_tensor.dtype}')

_int_tensor= torch.tensor([1,2,3,4])
print(_int_tensor.type())
# change type to float
_float_tensor=_int_tensor.type(torch.FloatTensor)
print(_float_tensor.type())

print(f'Size of tensor {_float_tensor.size()}')
print(f'Dimension of tensor {_float_tensor.ndimension()}')
```

#### Change shape of tensor

```python
# change from 1D to 2D
twoD_float_tensor=_float_tensor.view(4,1)
print(f'size {twoD_float_tensor.size()}')
print(f'Dimension {twoD_float_tensor.ndimension()}')

# for dynamic size tensor use -1 but can use in 1 arguement only
twoD_float_tensor=_float_tensor.view(-1,1)
print(twoD_float_tensor.size())
```

#### Numpy to tensor

```python
arr=np.array([1,2,3])
_tensor=torch.from_numpy(arr)
print(_tensor)
```

#### Tensor to numpy

```python
back_numpy = _tensor.numpy()
print(back_numpy)
```

Note both only _tensor point to same array

```python
arr[:] = 0
print(f'tensor {_tensor}')
print(f'array {back_numpy}')
```

#### Pandas series to tensor

```python
pandas_series = pd.Series([0.1,2,0.3,10.1])
new_tensor=torch.from_numpy(pandas_series.values)
print(new_tensor)
print(f'dtype {new_tensor.dtype}')
print(f'type {new_tensor.type()}')
```

#### access values in tensor as python number

```python
_tens=torch.tensor([0,1,2,3])
print(f'tensor {_tens}')

print(f'first item {_tens[0].item()} first tensor value {_tens[0]}')
print(type(_tens[0]))
print(type(_tens[0].item()))
```

### Tensor functions

#### mean

```python
math_tensor =torch.tensor([1.0,-1.0,1,-1])
print(math_tensor.mean())
```

standard deviation

```python
print(math_tensor.std())
```

max and min

```python
max_min_tensor = torch.tensor([1,1,3,5,5])
print(max_min_tensor.max())
print(max_min_tensor.min())
```

sine

```python
pi_tensor=torch.tensor([0,np.pi/2,np.pi])
sin = torch.sin(pi_tensor)
print("The sin result of pi tensor: ", sin)
```

Note this is a tensor. So you need torch.linspace

```python
pi_tensor=torch.linspace(0,2*np.pi,100)

sin_result= torch.sin(pi_tensor)
plt.plot(pi_tensor.numpy(),sin_result.numpy())
```

### Tensor Operations
- notice this is like vector operations

#### addition

```python
u = torch.tensor([1,0])
v = torch.tensor([0,1])
w=u+v
#w=u-v
print("The resultant tensor is",w)
```

Let's plot these vectors

```python
plotVec([
    {"vector": u.numpy(), "name": 'u', "color": 'r'},
    {"vector": v.numpy(), "name": 'v', "color":'b'},
    {"vector": w.numpy(), "name": 'w', "color": 'g'}
])
plt.show()
```

```python
#### tensor multiplication
v=2*u
print(v)

# dot product
print(torch.dot(u,v))
```
