```python
import torch 
import numpy as np 
import pandas as pd

import matplotlib.pyplot as plt
```

```python
# Plot vecotrs, please keep the parameters in the same length
# @param: Vectors = [{"vector": vector variable, "name": name of vector, "color": color of the vector on diagram}]
%matplotlib widget
    
def plotVec(vectors):
    ax = plt.axes()
    
    # For loop to draw the vectors
    for vec in vectors:
        ax.arrow(0, 0, *vec["vector"], head_width = 0.05,color = vec["color"], head_length = 0.1)
        plt.text(*(vec["vector"] + 0.1), vec["name"])
    
    plt.ylim(-2,2)
    plt.xlim(-2,2)
```

#### Create tensor
- type
- datatype
- setting datatype manually
- size and dimension 

```python
_list=[1,2,3,4]
_tensor=torch.tensor(_list)
print(_tensor)
print(f'Type: {_tensor.type()}')
print(f'Type: {_tensor.dtype}')

# create float tensor
_tensor=torch.tensor([1.0,4.5])
print(_tensor)
print(f'Type: {_tensor.type()}')
print(f'Type: {_tensor.dtype}')

_tensor=torch.tensor([1.0,4.5],dtype=torch.float64)
print(f'Type: {_tensor.dtype}')

_int_tensor= torch.tensor([1,2,3,4])
print(_int_tensor.type())
# change type to float
_float_tensor=_int_tensor.type(torch.FloatTensor)
print(_float_tensor.type())

print(f'Size of tensor {_float_tensor.size()}')
print(f'Dimension of tensor {_float_tensor.ndimension()}')
```

#### Change shape of tensor

```python
# change from 1D to 2D
twoD_float_tensor=_float_tensor.view(4,1)
print(f'size {twoD_float_tensor.size()}')
print(f'Dimension {twoD_float_tensor.ndimension()}')

# for dynamic size tensor use -1 but can use in 1 arguement only
twoD_float_tensor=_float_tensor.view(-1,1)
print(twoD_float_tensor.size())
```

### Tensor Operations

#### Derivatives

Let us create the tensor x and set the parameter requires_grad to true, means you need to take derivative of x
- Here we take derivative of y=x^2 and find it's value at x = 3

```python
# want to take derivative at x=2
x = torch.tensor(3.0, requires_grad=True)
print("The tensor x: ", x)

y=x**2
print(f'The function you want to take derivative of {y}')

y.backward()
print("The tensor x=2:",x.grad)

```

```python
print(f'data: {x.data}') # at x =3
print('grad_fn: {x.grad_fn}')
print('grad:', x.grad) # take derivative and put x=3
print("requires_grad",x.requires_grad)
```

```python
print(f'data: {y.data}') # y at x =3
print('grad_fn: {y.grad_fn}')
print('grad:', y.grad) # take derivative and put x=3
print("is_leaf",y.is_leaf)
print("requires_grad",y.requires_grad)
```
